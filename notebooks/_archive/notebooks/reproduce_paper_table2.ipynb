# Reproduce paper Table 2 (maps + scatters)

This notebook does **not** retrain the simplified `train_*.ipynb` models.
It calls `reproduce_paper_table2.py`, which loads the nn4sedratetoc Table-2
checkpoints and paper prediction maps so metrics match
`agujournaltemplate_2.tex`.

**Why:** seeding alone is not enough. Paper numbers need the exact
`masked_features` preprocess, `trainDNN`/`trainMTL` trainers, and saved
weights (`*_table2.pt`, attention best ckpts).


```python
from pathlib import Path
import runpy

script = Path("reproduce_paper_table2.py")
if not script.exists():
    script = Path().resolve() / "reproduce_paper_table2.py"
print("Running", script.resolve())
runpy.run_path(str(script), run_name="__main__")
```

```python
import pandas as pd
from pathlib import Path

out = Path().resolve().parents[0] / "data/output/predictionmaps_SedRate/notebooks_paper_table2"
if out.name != "notebooks_paper_table2":
    # if cwd is project root
    cand = Path("data/output/predictionmaps_SedRate/notebooks_paper_table2")
    out = cand if cand.exists() else out

csv = out / "table2_reproduction_metrics.csv"
print("metrics:", csv)
display(pd.read_csv(csv))
print("scatters:", out / "scatters")
print("maps:", out / "maps")
```
